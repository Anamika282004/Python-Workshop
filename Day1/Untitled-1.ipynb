```python
#array concept
import array as arr
a=arr.array('d',[1.1,2.1,3.1,4.1,5.1,6.1])
print(a[2])
print(a[1:4])
print(a[1:6:2])
print(a[-5])
print(a[-6:-2])
```

```python
#reverse an array
import array as arr
a=arr.array('d',[1.1,2.1,3.1,4.1,5.1,6.1])
print(a[::-1])

#changing array values
a[0]=10.1
print(a)
a[2:5]=arr.array('d',[30.1,40.1,50.1])
print(a)
```

```python
#append values
import array as num
num1=num.array('i',[1,2,3])
num1.append(4)
num1.append(5)
print(num1)
```
