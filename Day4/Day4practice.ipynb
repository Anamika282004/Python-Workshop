```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.get_dataset_names()
df = sns.load_dataset("tips")
df



```

```python
print(df.shape)
print(df.info())
print(df.describe())
print(df.head())
print(df.tail())
print(df.columns)
print(df.dtypes)
df.describe()
df['total_bill']
df[['total_bill','tip']]
df.iloc[5]
df.loc[df['sex'] =='Female']
df.loc[(df['day'] =='Sun') & (df['time']=='Dinner')]
```

```python
#data transformation
df['tips_pct']=df['tip']/df['total_bill']*100
df.head()
df['name']='Arka'
df.head()
df["isWeekend"]= df['day'].isin(['Sat','Sun'])
df.head()
df.isna().sum()
```

```python
#aggregation
df.groupby('day')['tip'].sum()
df.groupby('day')['tip'].mean()
df.groupby(['tip','time']).agg(
    t_mean=('tip','mean'),
    t_median=('tip','median'),
    b_mean=('total_bill','mean'),
    n=('tip','size')
)
```

```python
sns.histplot(df['tips_pct'])
```

```python
sns.boxplot(data=df,x='day',y='tips_pct')
```

```python
import matplotlib.pyplot as plt
plt.xlabel('Total Bill')
plt.ylabel('Tip %')
plt.title('Tip Percentage by Day of the Week')
sns.scatterplot(data=df,x='total_bill',y='tip',hue='day')
```
