```python
import pandas as pd
def Dynamic_exploraty_analysis():
    path_of_file = input("Enter Path of File: ")
    
    try:
        df = pd.read_csv(path_of_file)
    except Exception as exp:
        return {"error": str(exp)}
    
    return {
        "head": df.head(),
        "tail": df.tail(),
        "shape": df.shape,
        "columns": df.columns,
        "data_types": df.dtypes,
        "info": df.info(), 
        "describe": df.describe(),
        "missing_values": df.isnull().sum(),
        "duplicates": df.duplicated().sum(),
    }


Dynamic_exploraty_analysis()
```
