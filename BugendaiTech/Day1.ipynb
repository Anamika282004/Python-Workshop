```python
#Looping in lists
fruits = ["Apple","Banana","Guava"]
for x in fruits:
    print(x)

for i in range(len(fruits)):
    print(fruits[i])
i=0
while i<len(fruits):
    print(fruits[i])
    i+=1
```

```python
Salary = [50000, 26000, 70000, 60000, 40000, 90000]
ans = []
for s in Salary:
   if s>60000:
       ans.append(s)

print(ans)

#comprehension form
newans = [z for z in Salary if z>40000]
print(newans)

value = tuple(Salary)
print(value.index(70000))
print(value.count(50000))
```

```python
#nested dictionaries
student_data = {
    "Ram":{
        "age":30,
        "marks":95
    },
    "Sarita":{
        "age":40,
        "marks":56
    },
    "Anjum":{
        "age":23,
        "marks":87
    }
}
for x, y in student_data.items():
    print(x)
    for z in y:
        print(z," ",y[z])
```
