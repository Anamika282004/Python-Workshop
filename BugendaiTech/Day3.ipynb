```python
def get_match_count(city):
    mask = df['city']==city
    return df[mask]
```

```python
get_match_count('Bangalore')
```

```python
def Find_team(team):
    data1 = df['toss_winner']==team
    data2 = df['winner']==team
    return df[data1 & data2]
```

```python
filter = Find_team("Mumbai Indians")
# filter.shape[0]
filter
```

```python
#Gives 10 initial rows
def preview_head(df,n):
    return df.head(n)

preview_head(df,3)
```

```python
#last n rows
def preview_below_rows(df,n):
    return df.tail(n)

preview_below_rows(df,3)
```

```python
#only column names
def col_name(df):
    return df.columns   #df.columns gives cols    

col_name(df)
```

```python
#count missing values
def missing_count(df):
    return df.isnull().sum()
missing_count(df)
```

```python
#select columns
def select_cols(col):
    return df[col]
select_cols('season')

#multiple columns
def mul_cols(df,cols):
    return df[cols]

mul_cols(df,['team1','team2','winner'])
```

```python
#filteration
def filter_teams(df):
    mask1= df['city']=='Bangalore'
    mask2=df['toss_winner']=='Gujarat Lions'
    return df[mask1 & mask2]
filter_teams(df)
```

```python
#get datatypes
def data_type(df):
    return df.dtypes

data_type(df)
```

```python
def dynamic_eda(df):
    return {
        "shape": df.shape,
        "missing_values": df.isnull().sum(),
        "describe": df.describe()
    }

dynamic_eda(df)
```
