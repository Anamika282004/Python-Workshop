```python
#union of sets
set1={1,2,3,4}
set2={3,4,5,6}
set3=set1.union(set2)
print(set3)
set4={1,2,3}
set5={3,4,5}
set6=set3.union(set4,set5)
print(set6)
set7=set1|set2 # | operator can also be used for union
print(set7)
set8=set1.intersection(set2) #intersection of sets 
# & operator can also be used for intersection
# - operator can be used for difference of sets, A-B gives elements in A not in B
#symmetric difference can be found using ^ operator, A^B gives elements in A or B but not in both
#issubset can be checked using issubset() method, A.issubset(B) returns True if A is subset of B
# isdisjoint can be checked using isdisjoint() method, A.isdisjoint(B) returns True if A and B have no common elements
print(set8)
```

```python
# dictionary is a collection of key-value pairs, where each key is unique and maps to a value. It is mutable and unordered.
# mydict={}
# mydict={1:'apple',2:ball}
# mydict= dict({1:'apple',2:'ball'})
mydict={'name':'Alice','age':30,'city':'New York'}
print(mydict['name'])
mydict['address']='Kankarbagh,Patna'
print(mydict['address'])
```

```python
#printing key and values of a dictionary
student={'name':'Anamika','age':20, 'marks':90}
for i in student:
    print(i)
#for ascessing values using keys
for i in student:
    print(student[i])
#for ascessing key and values using items() method
for i,j in student.items():
    print(i,j)
```

```python
#nested dictionary
students = {
    "101": {
        "name": "Anamika",
        "age": 20,
        "marks": 98
    },
    "102": {
        "name": "Kritika",
        "age": 22,
        "marks": 95
    }
}
for i,j in students.items():
    print("id:",i)
    print("name",j['name'])
    print("age",j['age'])
    print("marks",j['marks'])
```

```python
studentsMarks = {
    "Anamika":98,
    "Kritika":95,
    "Anashika":90,
    "Rahul":85,
    "Kaushik":75
}
for i,j in studentsMarks.items():
    if j>=80:
        print(i,"has passed")
    else:
        print(i,"has failed")
```
